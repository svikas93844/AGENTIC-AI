# 🏥 Multi-Agent Healthcare Monitor System
### Agentic AI Framework Demonstrating 4 Core Multi-Agent Architectures:
1. **Peer-to-Peer (P2P) Architecture**: Direct communication between medical specialist agents without central bottlenecks.
2. **Blackboard Architecture**: Central shared memory board updated asynchronously by sensor & knowledge source agents.
3. **Parallel Execution Architecture**: Concurrent multi-channel vital processing for low-latency anomaly detection.
4. **Sequential / Pipeline Architecture**: Multi-stage linear pipeline (`Ingestion -> Feature Extraction -> ML Anomaly Scoring -> Triage -> Alert Dispatch`).

## 2. Dataset Generation & Machine Learning Model Training
We generate a synthetic multi-parameter patient vital dataset containing **5,000 patient samples** with vital sign telemetry:
- Heart Rate (bpm)
- SpO2 Oxygen Saturation (%)
- Systolic & Diastolic BP (mmHg)
- Respiration Rate (breaths/min)
- Body Temperature (°C)
- Blood Glucose (mg/dL)
- Troponin Level (ng/mL)
- Patient Age

**Target Clinical Classes**:
- `0`: Normal Baseline
- `1`: Mild Distress
- `2`: Urgent ICU Care
- `3`: Critical Emergency (Cardiac / Septic Shock)

```python
def generate_healthcare_dataset(n_samples: int = 5000, random_seed: int = 42) -> pd.DataFrame:
    np.random.seed(random_seed)
    n_normal = int(n_samples * 0.50)
    n_mild = int(n_samples * 0.25)
    n_urgent = int(n_samples * 0.15)
    n_critical = n_samples - (n_normal + n_mild + n_urgent)
    
    def gen(c, hr_m, hr_s, spo2_m, spo2_s, sbp_m, sbp_s, dbp_m, dbp_s, rr_m, rr_s, temp_m, temp_s, gluc_m, gluc_s, trop_m, trop_s, label):
        return pd.DataFrame({
            'heart_rate': np.clip(np.random.normal(hr_m, hr_s, c), 40, 210),
            'spo2': np.clip(np.random.normal(spo2_m, spo2_s, c), 70, 100),
            'systolic_bp': np.clip(np.random.normal(sbp_m, sbp_s, c), 70, 220),
            'diastolic_bp': np.clip(np.random.normal(dbp_m, dbp_s, c), 40, 130),
            'respiration_rate': np.clip(np.random.normal(rr_m, rr_s, c), 8, 45),
            'body_temp': np.clip(np.random.normal(temp_m, temp_s, c), 35.0, 42.0),
            'blood_glucose': np.clip(np.random.normal(gluc_m, gluc_s, c), 50, 450),
            'troponin_level': np.clip(np.random.normal(trop_m, trop_s, c), 0.0, 15.0),
            'age': np.random.randint(18, 90, c),
            'risk_level': label
        })
    
    df_n = gen(n_normal, 72, 8, 98.5, 1.0, 120, 8, 78, 6, 16, 2, 36.8, 0.3, 95, 12, 0.01, 0.005, 0)
    df_m = gen(n_mild, 95, 12, 95.0, 2.0, 138, 12, 88, 8, 22, 3, 38.2, 0.6, 140, 25, 0.03, 0.01, 1)
    df_u = gen(n_urgent, 125, 15, 90.0, 3.5, 165, 18, 102, 12, 28, 4, 39.1, 0.8, 220, 50, 0.15, 0.08, 2)
    df_c = gen(n_critical, 155, 22, 82.0, 5.0, 75, 15, 45, 10, 36, 6, 40.2, 1.2, 340, 80, 2.50, 1.20, 3)
    
    df = pd.concat([df_n, df_m, df_u, df_c], ignore_index=True)
    return df.sample(frac=1.0, random_state=random_seed).reset_index(drop=True)

dataset = generate_healthcare_dataset(5000)
print("Dataset Shape:", dataset.shape)
dataset.head()
```

### Training Random Forest Triage Classifier
Following strict ML featurization ordering, we split the data **before** scaling.

```python
feature_cols = ['heart_rate', 'spo2', 'systolic_bp', 'diastolic_bp', 'respiration_rate', 'body_temp', 'blood_glucose', 'troponin_level', 'age']
X = dataset[feature_cols]
y = dataset['risk_level']

# Split first
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

# Scale training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train Random Forest
rf_model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
rf_model.fit(X_train_scaled, y_train)

# Evaluate
y_pred = rf_model.predict(X_test_scaled)
acc = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {acc * 100:.2f}%")

# Plot Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Mild', 'Urgent', 'Critical'], yticklabels=['Normal', 'Mild', 'Urgent', 'Critical'])
plt.title("Confusion Matrix - Healthcare Risk Classifier")
plt.xlabel("Predicted Risk")
plt.ylabel("Actual Risk")
plt.show()
```

### Feature Importance Analysis
Let's visualize which vital sign indicators have the highest predictive weight in classifying clinical distress.

```python
importance_df = pd.DataFrame({
    'Feature': feature_cols,
    'Importance': rf_model.feature_importances_
}).sort_values(by='Importance', ascending=False)

plt.figure(figsize=(9, 4))
sns.barplot(x='Importance', y='Feature', data=importance_df, palette='viridis')
plt.title("Feature Importances for Multi-Agent Healthcare Monitor")
plt.show()
```

## 3. Architecture 1: Peer-to-Peer (P2P) Multi-Agent Architecture
In P2P architecture, specialist medical agents (**Cardiologist**, **Pulmonologist**, **Endocrinologist**) exchange message objects directly over peer queues without a central supervisor node.

```python
@dataclass
class P2PMessage:
    sender: str
    recipient: str
    content: str
    timestamp: float = field(default_factory=time.time)

class SpecialistAgentP2P:
    def __init__(self, name: str, specialty: str):
        self.name = name
        self.specialty = specialty
        self.inbox: List[P2PMessage] = []
        self.peers: Dict[str, 'SpecialistAgentP2P'] = {}

    def connect_peer(self, peer: 'SpecialistAgentP2P'):
        self.peers[peer.name] = peer

    def send_message(self, recipient_name: str, content: str):
        if recipient_name in self.peers:
            msg = P2PMessage(sender=self.name, recipient=recipient_name, content=content)
            self.peers[recipient_name].inbox.append(msg)

    def evaluate(self, vitals: Dict[str, float]) -> List[str]:
        logs = []
        if self.specialty == 'Pulmonology' and vitals['spo2'] < 90:
            logs.append(f"[{self.name}] SpO2={vitals['spo2']}% (Severe Hypoxia). Messaging Cardiologist.")
            self.send_message('CardiologistAgent', f"Hypoxia alert SpO2={vitals['spo2']}%. Evaluate cardiac load.")
        elif self.specialty == 'Cardiology' and vitals['heart_rate'] > 120:
            logs.append(f"[{self.name}] HR={vitals['heart_rate']} bpm (Severe Tachycardia). Messaging Endocrinologist.")
            self.send_message('EndocrinologistAgent', f"Tachycardia HR={vitals['heart_rate']} bpm. Check blood glucose.")
        
        while self.inbox:
            msg = self.inbox.pop(0)
            logs.append(f"[{self.name}] Received P2P message from {msg.sender}: '{msg.content}'")
            if 'Hypoxia' in msg.content and self.specialty == 'Cardiology':
                logs.append(f"[{self.name}] P2P Response -> Heart Rate: {vitals['heart_rate']} bpm. Ischemic heart strain elevated!")
        return logs

# Demonstration
cardio = SpecialistAgentP2P("CardiologistAgent", "Cardiology")
pulmo = SpecialistAgentP2P("PulmonologistAgent", "Pulmonology")
endo = SpecialistAgentP2P("EndocrinologistAgent", "Endocrinology")

cardio.connect_peer(pulmo)
cardio.connect_peer(endo)
pulmo.connect_peer(cardio)
endo.connect_peer(cardio)

sample_vitals = {'heart_rate': 140, 'spo2': 85, 'troponin_level': 1.2, 'blood_glucose': 260}
print("=== P2P Execution Log ===")
for log in pulmo.evaluate(sample_vitals) + cardio.evaluate(sample_vitals) + endo.evaluate(sample_vitals):
    print(log)
```

## 4. Architecture 2: Blackboard Architecture
In Blackboard architecture, a central shared memory (`Blackboard`) stores facts & hypotheses. Independent **Knowledge Source (KS)** agents asynchronously read and post hypotheses.

```python
class Blackboard:
    def __init__(self):
        self.entries = []
        self.alert_status = "GREEN"

    def post(self, agent_name: str, category: str, payload: Any):
        entry = {'agent': agent_name, 'category': category, 'payload': payload}
        self.entries.append(entry)
        if category == "CRITICAL_ALERT":
            self.alert_status = "RED"

bb = Blackboard()
bb.post("TelemetryIngestionKS", "VITAL_READING", sample_vitals)

# Cardiac Knowledge Source
if sample_vitals['troponin_level'] > 0.1 or sample_vitals['heart_rate'] > 130:
    bb.post("CardiacEvaluatorKS", "CRITICAL_ALERT", f"Myocardial Ischemia Risk! Troponin: {sample_vitals['troponin_level']}")

print("=== Blackboard State ===")
print("Global Alert Status:", bb.alert_status)
for entry in bb.entries:
    print(f"  [{entry['agent']} | {entry['category']}]: {entry['payload']}")
```

## 5. Architecture 3: Parallel Execution Architecture
Parallel architecture runs dedicated worker agents concurrently across CPU/async threads for ultra-fast multi-sensor telemetry processing.

```python
async def worker_cardio(vitals):
    await asyncio.sleep(0.01)
    return {'sensor': 'Cardiovascular', 'status': 'HIGH_RISK' if vitals['heart_rate'] > 120 else 'NORMAL'}

async def worker_respiratory(vitals):
    await asyncio.sleep(0.01)
    return {'sensor': 'Respiratory', 'status': 'HIGH_RISK' if vitals['spo2'] < 90 else 'NORMAL'}

async def worker_metabolic(vitals):
    await asyncio.sleep(0.01)
    return {'sensor': 'Metabolic', 'status': 'ELEVATED' if vitals['blood_glucose'] > 200 else 'NORMAL'}

async def run_parallel_simulation(vitals):
    t0 = time.perf_counter()
    results = await asyncio.gather(
        worker_cardio(vitals),
        worker_respiratory(vitals),
        worker_metabolic(vitals)
    )
    elapsed = (time.perf_counter() - t0) * 1000
    print(f"=== Parallel Execution Completed in {elapsed:.2f} ms ===")
    for res in results:
        print(f"  Sensor Worker [{res['sensor']}]: Status = {res['status']}")

await run_parallel_simulation(sample_vitals)
```

## 6. Architecture 4: Sequential / Pipeline Architecture
A structured linear pipeline that processes data through deterministic sequential stages.

```python
def pipeline_execution(vitals, model, scaler):
    # Stage 1: Ingestion
    s1_out = {'vitals': vitals, 'ingest_status': 'VALIDATED'}
    
    # Stage 2: Feature Engineering
    shock_index = vitals['heart_rate'] / max(vitals.get('systolic_bp', 120), 1)
    s2_out = {**s1_out, 'shock_index': round(shock_index, 2)}
    
    # Stage 3: ML Inference
    feats = np.array([[vitals['heart_rate'], vitals['spo2'], 120, 80, 20, 37.0, vitals['blood_glucose'], vitals['troponin_level'], 60]])
    feats_scaled = scaler.transform(feats)
    pred_code = int(model.predict(feats_scaled)[0])
    risk_label = {0: 'Normal', 1: 'Mild Distress', 2: 'Urgent Care', 3: 'Critical Emergency'}[pred_code]
    s3_out = {**s2_out, 'risk_label': risk_label}
    
    # Stage 4: Clinical Triage & Alert Dispatch
    action = "ACTIVATE RAPID RESPONSE CODE RED" if pred_code == 3 else "ROUTINE MONITORING"
    s4_out = {**s3_out, 'action': action}
    return s4_out

pipeline_result = pipeline_execution(sample_vitals, rf_model, scaler)
print("=== Sequential Pipeline Output ===")
for k, v in pipeline_result.items():
    print(f"  {k}: {v}")
```
